In Pandas, pd.crosstab() is used to compute a cross-tabulation of two (or more) factors. It shows the frequency distribution of values as a matrix, making it great for analyzing relationships between categorical variables.


This command generates a cross-tabulation table (contingency table) that shows the frequency distribution of Grade values within each Department using pandas.crosstab().

```python
pd.crosstab(stud_df['Department'], stud_df['Grade'])
```

**Crosstab + Total Students**


```python
# Normalize grades to uppercase (if needed)
stud_df['Grade'] = stud_df['Grade'].str.upper()
```

```python
# Generate crosstab of Department vs. Grade
dept_grade_crosstab = pd.crosstab(stud_df['Department'], stud_df['Grade'])


# Add total students per department
dept_grade_crosstab['Total_Students'] = dept_grade_crosstab.sum(axis=1)



# Display the result
print(dept_grade_crosstab)
```

```python
# Generate crosstab of Department vs. Grade
dept_grade_crosstab = pd.crosstab(stud_df['Department'], stud_df['Grade'])


# Add total students per grade (column-wise)

dept_grade_crosstab.loc['Total_Grades'] = dept_grade_crosstab.sum(axis=0)

# Display the result
print(dept_grade_crosstab)
```

The groupby() function in Pandas is a powerful tool used to group data based on one or more columns, and then apply aggregation functions (like sum(), mean(), count(), etc.) to each group.


This command counts the number of students (Student_ID) for each combination of Department and Grade in the DataFrame stud_df.

```python
stud_df.groupby(['Department', 'Grade']).count()['Student_ID']
```
