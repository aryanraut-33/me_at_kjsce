```python
def encode_categorical(df):
    for col in df.select_dtypes(include='object').columns:
        if df[col].nunique() == 2:
            unique_vals = df[col].unique()
            df[col] = df[col].map({unique_vals[0]: 0, unique_vals[1]: 1})
    return df

df_encoded = encode_categorical(df)

print("Categorical column [Sex] encoded successfully!")
df_encoded.head()

df['Sex'].head(10)
```

```python
df = pd.read_csv("./dataset/titanic.csv")

def normalize_data(df):
    print("Normalizing columns:")
    for col in df.select_dtypes(include=['int64', 'float64']).columns:
        if col == 'Age' or col == 'Fare':  # normalize only Age and Fare
            min_val = df[col].min()
            max_val = df[col].max()
            if min_val != max_val:
                df[col] = (df[col] - min_val) / (max_val - min_val)
                print(f"- {col}")
    return df
```

```python
df_normalize = normalize_data(df)
df_normalize['Age'].head(5)
```

```python
df = pd.read_csv("./dataset/titanic.csv")

def discretize_data(df):
    print("Discretizing the following continuous columns:\n")
    if 'Age' in df.columns:
        # Divide Age into 4 bins (Child, Teen, Adult, Senior)
        df['AgeGroup'] = pd.cut(df['Age'], bins=[0, 12, 19, 50, 80], 
                                labels=['Child', 'Teen', 'Adult', 'Senior'])
        print("- Age: AgeGroup (Child/Teen/Adult/Senior)")
    
    print("\nDiscretization complete.\n")
    return df
```

```python
df = discretize_data(df)
df.head(25)
```
